```python
import os
from pathlib import Path

import pandas as pd

from utils import find_project_root
```

```python
data_dir = find_project_root() / 'data'
```

```python
bacteria = pd.read_csv(data_dir / "samplesheet_bacteria.csv")
bacteria.head()
```

```python
# Create the target directory if it doesn't exist
target_dir = data_dir / 'bacteria'
target_dir.mkdir(parents=True, exist_ok=True)

# Iterate over the rows of the dataframe and create symlinks
for index, row in bacteria.iterrows():
    forward_path = Path(find_project_root() / row['forwardReads'])
    reverse_path = Path(find_project_root() / row['reverseReads'])

    # Create symlinks for forward reads
    forward_symlink = target_dir / forward_path.name
    if not forward_symlink.exists():
        os.symlink(forward_path.resolve(), forward_symlink)

    # Create symlinks for reverse reads
    reverse_symlink = target_dir / reverse_path.name
    if not reverse_symlink.exists():
        os.symlink(reverse_path.resolve(), reverse_symlink)
```
